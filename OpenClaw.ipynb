```python
import nest_asyncio
import torch
from transformers import AutoModelForCausalLM, AutoTokenizer
from telegram import Update
from telegram.ext import ApplicationBuilder, CommandHandler, MessageHandler, filters, ContextTypes

# Mengizinkan asyncio berjalan di dalam Google Colab
nest_asyncio.apply()

# Konfigurasi Token dan Model
TOKEN = "7296739099:AAFuZ6dIxXmsC1nrBdg916YiWj1EtCFcYEI"
MODEL_NAME = "Qwen/Qwen2-0.5B-Instruct"

# --- [1/3] MEMUAT MODEL AI KE MEMORI ---
print("--- MEMUAT MODEL AI KE GPU ---")
device = "cuda" if torch.cuda.is_available() else "cpu"

tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    dtype=torch.float16 if device == "cuda" else torch.float32,
    device_map="auto" if device == "cuda" else None
)
print("-> Model AI berhasil dimuat!")

# --- [2/3] DEFINISI HANDLER BOT TELEGRAM ---

# Perintah /start
async def start_command(update: Update, context: ContextTypes.DEFAULT_TYPE):
    user_name = update.effective_user.first_name
    welcome_text = (
        f"Halo {user_name}! 👋\n"
        "Saya adalah AI Chatbot berbasis Qwen2. "
        "Kirimkan pesan atau pertanyaan apa saja, saya siap membantu!"
    )
    await update.message.reply_text(welcome_text)

# Memproses Pesan Masuk dari User
async def handle_message(update: Update, context: ContextTypes.DEFAULT_TYPE):
    user_text = update.message.text

    # Menampilkan indikator "typing..." di Telegram
    await context.bot.send_chat_action(chat_id=update.effective_chat.id, action="typing")

    # Format percakapan standar Qwen
    messages = [
        {"role": "system", "content": "Kamu adalah asisten AI yang ramah, serba tahu, dan membantu."},
        {"role": "user", "content": user_text}
    ]
    prompt = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)

    # Generasi Jawaban oleh AI
    inputs = tokenizer([prompt], return_tensors="pt").to(device)
    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=512,
            do_sample=True,
            temperature=0.7,
            top_p=0.9
        )

    # Mengambil teks balasan saja
    generated_ids = [
        output_ids[len(input_ids):] for input_ids, output_ids in zip(inputs.input_ids, outputs)
    ]
    response = tokenizer.batch_decode(generated_ids, skip_special_tokens=True)[0]

    # Mengirim balasan ke user Telegram
    await update.message.reply_text(response)

# --- [3/3] MENJALANKAN BOT ---
if __name__ == "__main__":
    app = ApplicationBuilder().token(TOKEN).build()

    # Registrasi Handler
    app.add_handler(CommandHandler("start", start_command))
    app.add_handler(MessageHandler(filters.TEXT & ~filters.COMMAND, handle_message))

    print("\n==========================================")
    print("🤖 BOT TELEGRAM AKTIF & SIAP DICHAT!")
    print("==========================================")

    # Menjalankan bot (drop_pending_updates=True agar bot tidak memproses pesan lama)
    app.run_polling(drop_pending_updates=True)
```

```python
# 1. Install semua pustaka yang dibutuhkan
!pip install nest_asyncio python-telegram-bot transformers accelerate torch bitsandbytes ddgs --quiet

import os
import sys
import time
import sqlite3
import warnings
from datetime import datetime, timezone

# --- [1/6] PATCH SUPAYA TIDAK SPAM WARNING JUPYTER CLIENT ---
try:
    import jupyter_client.session
    jupyter_client.session.utcnow = lambda: datetime.now(timezone.utc)
except Exception:
    pass

os.environ["PYTHONWARNINGS"] = "ignore"
warnings.filterwarnings("ignore")
warnings.simplefilter("ignore")

import torch
import nest_asyncio
from duckduckgo_search import DDGS
from transformers import AutoModelForCausalLM, AutoTokenizer, BitsAndBytesConfig
from telegram import Update
from telegram.ext import ApplicationBuilder, CommandHandler, MessageHandler, filters, ContextTypes

nest_asyncio.apply()

# --- KONFIGURASI BOT ---
BOT_TOKEN = "7296739099:AAFuZ6dIxXmsC1nrBdg916YiWj1EtCFcYEI"
MODEL_NAME = "Qwen/Qwen2.5-7B-Instruct"

# --- [2/6] DATABASE MEMORI (SQLite + Proteksi Corrupt) ---
def init_db():
    db_path = "agent_memory.db"
    try:
        conn = sqlite3.connect(db_path)
        cursor = conn.cursor()
        cursor.execute("""
            CREATE TABLE IF NOT EXISTS chat_history (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                user_id INTEGER,
                role TEXT,
                content TEXT
            )
        """)
        conn.commit()
        conn.close()
    except sqlite3.DatabaseError:
        print(f"[{time.strftime('%H:%M:%S')}] ⚠️ Database corrupt terdeteksi! Reset otomatis {db_path}...")
        if os.path.exists(db_path):
            os.remove(db_path)
        conn = sqlite3.connect(db_path)
        cursor = conn.cursor()
        cursor.execute("""
            CREATE TABLE IF NOT EXISTS chat_history (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                user_id INTEGER,
                role TEXT,
                content TEXT
            )
        """)
        conn.commit()
        conn.close()

def save_memory(user_id, role, content):
    conn = sqlite3.connect("agent_memory.db")
    cursor = conn.cursor()
    cursor.execute("INSERT INTO chat_history (user_id, role, content) VALUES (?, ?, ?)", (user_id, role, content))
    conn.commit()
    conn.close()

def get_recent_memory(user_id, limit=4):
    conn = sqlite3.connect("agent_memory.db")
    cursor = conn.cursor()
    cursor.execute("SELECT role, content FROM chat_history WHERE user_id = ? ORDER BY id DESC LIMIT ?", (user_id, limit))
    rows = cursor.fetchall()
    conn.close()
    return list(reversed(rows))

init_db()

# --- [3/6] FUNGSI PENCARIAN WEB ---
def search_web(query, max_results=3):
    try:
        results = list(DDGS().text(query, max_results=max_results))
        if not results:
            return "Tidak ditemukan hasil pencarian relevan."
        snippets = [f"[{i+1}] {r['title']}: {r['body']}" for i, r in enumerate(results)]
        return "\n".join(snippets)
    except Exception as e:
        return f"Pencarian web gagal: {e}"

# --- [4/6] MEMUAT MODEL AI ---
print("==========================================")
print("-> INISIALISASI SISTEM & MODEL AI")
print("==========================================")

device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"[{time.strftime('%H:%M:%S')}] 🖥️ Perangkat Aktif: {device.upper()}")

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_use_double_quant=True
) if device == "cuda" else None

print(f"[{time.strftime('%H:%M:%S')}] 🔄 Memuat Model '{MODEL_NAME}' ke VRAM...")
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    quantization_config=bnb_config,
    device_map="auto" if device == "cuda" else None,
    torch_dtype=torch.float16 if device == "cuda" else torch.float32
)
print(f"[{time.strftime('%H:%M:%S')}] ✅ AI Agent Siap Digunakan!")

# --- [5/6] LOGIKA BOT & LOG TERMINAL REAL-TIME ---
async def start_command(update: Update, context: ContextTypes.DEFAULT_TYPE):
    await update.message.reply_text("Halo! Bot aktif dan siap melayani.")

async def handle_message(update: Update, context: ContextTypes.DEFAULT_TYPE):
    user_id = update.effective_user.id
    user_text = update.message.text
    start_time = time.time()

    print(f"\n--------------------------------------------------")
    print(f"[{time.strftime('%H:%M:%S')}] 📩 Pesan Masuk ({user_id}): \"{user_text}\"")

    await context.bot.send_chat_action(chat_id=update.effective_chat.id, action="typing")

    # Bypass search jika permintaan adalah coding / membuat game
    skip_keywords = ["buatkan", "bikin", "code", "game", "script", "coding", "buat", "papan", "ular"]
    should_skip_search = any(kw in user_text.lower() for kw in skip_keywords)

    if should_skip_search:
        print(f"[{time.strftime('%H:%M:%S')}] ⏩ Perintah coding/kreatif terdeteksi -> Melewati Web Search.")
        search_context = ""
    else:
        print(f"[{time.strftime('%H:%M:%S')}] 🔍 Mencari informasi terbaru di Web Search...")
        search_context = search_web(user_text)
        print(f"[{time.strftime('%H:%M:%S')}] 🌐 Web Search selesai.")

    print(f"[{time.strftime('%H:%M:%S')}] 💾 Memuat riwayat chat dari SQLite...")
    past_memory = get_recent_memory(user_id)

    system_prompt = (
        "Kamu adalah AI Agent cerdas yang jujur, akurat, dan berbasis data nyata. "
        "Gunakan HASIL PENCARIAN INTERNET berikut sebagai acuan utama jawabanmu jika ada.\n\n"
        f"--- HASIL PENCARIAN INTERNET REAL-TIME ---\n{search_context}\n------------------------------------------"
    )

    messages = [{"role": "system", "content": system_prompt}]
    for role, content in past_memory:
        messages.append({"role": role, "content": content})
    messages.append({"role": "user", "content": user_text})

    print(f"[{time.strftime('%H:%M:%S')}] 🧠 Model Qwen 2.5 sedang memproses & menyusun respon...")
    prompt = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    inputs = tokenizer([prompt], return_tensors="pt").to(device)

    generated_ids = model.generate(
        inputs.input_ids,
        max_new_tokens=1512,
        do_sample=True,
        temperature=1,
        top_p=1.9
    )

    generated_ids = [output_ids[len(input_ids):] for input_ids, output_ids in zip(inputs.input_ids, generated_ids)]
    bot_response = tokenizer.batch_decode(generated_ids, skip_special_tokens=True)[0]

    save_memory(user_id, "user", user_text)
    save_memory(user_id, "assistant", bot_response)

    elapsed = time.time() - start_time
    print(f"[{time.strftime('%H:%M:%S')}] 📤 Respon berhasil dikirim ke Telegram! (Total Waktu: {elapsed:.2f} detik)")
    print(f"--------------------------------------------------")

    await update.message.reply_text(bot_response)

# --- [6/6] MENJALANKAN BOT ---
if __name__ == "__main__":
    app = ApplicationBuilder().token(BOT_TOKEN).build()
    app.add_handler(CommandHandler("start", start_command))
    app.add_handler(MessageHandler(filters.TEXT & ~filters.COMMAND, handle_message))

    print("\n==========================================")
    print("🚀 AI AGENT TELEGRAM AKTIF!")
    print("   Pantau proses aktivitas agent di bawah ini:")
    print("==========================================\n")

    app.run_polling(drop_pending_updates=True)
```
